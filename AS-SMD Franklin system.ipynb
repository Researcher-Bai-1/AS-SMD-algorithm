```python
import math
import random
import matplotlib.pyplot as plt
import time
import numpy as np

class Franklin_system():
    def __init__(self,N):
        self.N = N
        self.Set = [[-1,0],[-1,1]]
        self.G = np.zeros((2**(self.N+1)+1,2**(self.N+1)+1))
        for n in range(self.N+1):
            for k in range(1,2**n+1):
                self.Set.append([n,k])
                
        
    def hat_function(self,n,k,x):
        if n == -1:
            a0 = 1
            if k == 0:
                y = 1
            elif k == 1:
                y = x
        else:
            a0 = 2**n
            a1 = 1/a0
            a = (k-1)*a1
            b = k*a1
            if x <= a:
                y = 0
            elif x <= ((2*k-1)*a1/2):
                y = x-a
            elif x <= b:
                y = b-x
            else:
                y = 0
        return y*np.sqrt(a0)
        
    def franklin_system_cal(self,num,x):#num is real position of basis for num=1,2,3,4,dots
        Value_basis = []
        for A in self.Set[:num]:
            Value_basis.append(self.hat_function(A[0],A[1],x))
        return np.dot(np.array(Value_basis),self.COE[num-1])
            
    
    def int_hat_function(self,a,b,c,d,alpha,beta):
        return a*c*((beta**3/3-(b+d)*(beta**2)/2+b*d*beta)-(alpha**3/3-(b+d)*(alpha**2)/2+b*d*alpha))

    def int_0_hat_function(self,a,b,alpha,beta):
        y = a*((beta**2/2-b*beta)-(alpha**2/2-b*alpha))
        return y
    
    def inner_production_function(self,k,n,k1,n1):#2**n+k smaller than 2**n1+k1
        a = 2**n
        a1 = 2**n1
        S = 0
        lower_nk = (k-1)/a
        lower_n1k1 = (k1-1)/a1
        mid_nk = (2*k-1)/(a*2)
        mid_n1k1 = (2*k1-1)/(a1*2)
        upper_nk = k/a
        upper_n1k1 = k1/a1
        #print(n,k,n1,k1,'lower',lower_nk,lower_n1k1,'mid',mid_nk, mid_n1k1,'upper',upper_nk,upper_n1k1)
        if n>n1:
            if (lower_nk + 10**(-10) >= lower_n1k1) and (mid_nk <= 10**(-10) + mid_n1k1):
                S += self.int_hat_function(np.sqrt(a),lower_nk,np.sqrt(a1),lower_n1k1,lower_nk,mid_nk)
            elif lower_nk + 10**(-10) >= mid_n1k1 and mid_nk <= 10**(-10)+upper_n1k1:
                S += self.int_hat_function(np.sqrt(a),lower_nk,-np.sqrt(a1),upper_n1k1,lower_nk,mid_nk)
                #print('bb',n,k,n1,k1,S)
            if mid_nk + 10**(-10) >= lower_n1k1 and upper_nk <= 10**(-10) + mid_n1k1:
                S += self.int_hat_function(-np.sqrt(a),upper_nk,np.sqrt(a1),lower_n1k1,mid_nk,upper_nk)
                #print('cc',n,k,n1,k1,S)
            elif mid_nk  >= mid_n1k1 + 10**(-10) and upper_nk <= 10**(-10) + upper_n1k1:
                S += self.int_hat_function(-np.sqrt(a),upper_nk,-np.sqrt(a1),upper_n1k1,mid_nk,upper_nk)
                #print('dd',n,k,n1,k1,S)
        elif n == n1:
            if mid_nk <= 10**(-10) + upper_n1k1:
                S += self.int_hat_function(np.sqrt(a),lower_nk,-np.sqrt(a1),upper_n1k1,lower_nk,mid_nk)
        return S
    
    def schimidt_orthogonalization(self):
        self.G[0][0] = 1
        self.G[1][0] = 1/2
        self.G[0][1] = self.G[1][0]
        self.G[1][1] = 1/3
        for n in range(self.N+1):
            n_2 = 2**n
            for k in range(1,n_2+1):
                self.G[n_2+k][0] = self.int_0_hat_function(np.sqrt(n_2),(k-1)/n_2,(k-1)/n_2,(2*k-1)/(2*n_2))+self.int_0_hat_function(-np.sqrt(n_2),k/n_2,(2*k-1)/(2*n_2),k/n_2)
                self.G[n_2+k][1] = self.int_hat_function(np.sqrt(n_2),(k-1)/n_2,1,0,(k-1)/n_2,(2*k-1)/(2*n_2))+self.int_hat_function(-np.sqrt(n_2),k/n_2,1,0,(2*k-1)/(2*n_2),k/n_2)
                self.G[1][n_2+k] = self.G[n_2+k][1]
                self.G[0][n_2+k] = self.G[n_2+k][0]
                self.G[n_2+k][n_2+k] = self.int_hat_function(np.sqrt(n_2),(k-1)/n_2,np.sqrt(n_2),(k-1)/n_2,(k-1)/n_2,(2*k-1)/(2*n_2))+self.int_hat_function(-np.sqrt(n_2),k/n_2,-np.sqrt(n_2),k/n_2,(2*k-1)/(2*n_2),k/n_2)
                for n1 in range(n+1):
                    n1_2 = 2**n1
                    if n1 < n:
                        for k1 in range(1,n1_2+1):
                            self.G[n_2+k][n1_2+k1] = self.inner_production_function(k,n,k1,n1)
                            self.G[n1_2+k1][n_2+k] = self.G[n_2+k][n1_2+k1]
                    elif n1 == n:
                        for k1 in range(1,k):
                            self.G[n_2+k][n1_2+k1] = self.inner_production_function(k,n,k1,n1)
                            self.G[n1_2+k1][n_2+k] = self.G[n_2+k][n1_2+k1]
                                
        #print(self.G)
        P = np.linalg.cholesky(self.G)
        P = np.linalg.solve(P,np.eye(2**(self.N+1)+1))
        self.P = P
        self.COE = [ P[i][:i+1] for i in range(len(P))]
        
    def active_hat_values(self, L, x):
        indices = []
        values = []

        if L >= 1:
            indices.append(0)
            values.append(1.0)

        if L >= 2:
            indices.append(1)
            values.append(x)

        n = 0
        while 2**n + 1 < L and n <= self.N:
            a = 2**n
            if x >= 1.0:
                k = a
            else:
                k = int(np.floor(a * x)) + 1
            idx = a + k
            if idx < L:
                val = self.hat_function(n, k, x)
                if val != 0:
                    indices.append(idx)
                    values.append(val)
            n += 1
        return indices, values

    def franklin_system_values_fast(self, L, x):
        indices, values = self.active_hat_values(L, x)
        out = np.zeros(L)
        for idx, val in zip(indices, values):
            out[idx:L] += self.P[idx:L, idx] * val
        return out
```

```python
import math
import random
import matplotlib.pyplot as plt
import time
import numpy as np

class Mirror_descent_Banach():
    def __init__(self,noise,para,s,p,q,theta,tau,gamma0,N):
        self.s = s
        self.p = p
        self.pp = p/(p-1)
        self.q = q
        self.Ln = 1
        self.fn = np.zeros(1)
        self.f_average_n = np.zeros(1)
        self.theta = theta
        self.gamma0 = gamma0 #Initial step size
        self.tau = tau #Step decay rate
        self.Num = np.zeros(1)
        self.epoch = 1 #  Current number of iterations
        self.Hn = 0
        self.a0 = 2*np.cos(1.5*math.pi*np.arccos(1-abs(0-0.5))) + (2*0)**2 - 3*abs(0-1/4)**(2/3) + 3*np.cos(5*0) - np.exp(3*0) 
        self.a1 = 2*np.cos(1.5*math.pi*np.arccos(1-abs(1-0.5))) + (2*1)**2 - 3*abs(1-1/4)**(2/3) + 3*np.cos(5*1) - np.exp(3*1) 
        self.N = N
        self.noise = noise
        self.para = para
    
    def cal_basis_function(self,X):
        Output = franklin_system.franklin_system_values_fast(self.Ln, X)
        return Output
    
    def fn_cal(self,X):
        Output = self.cal_basis_function(X)
        return np.sum(self.fn*Output)
        
    def f_average_n_cal(self,X):
        Output = self.cal_basis_function(X)
        return np.sum(self.f_average_n*Output)

    def data_noise(self):
        c = self.para
        W = np.random.exponential()
        U = np.random.uniform(-math.pi/2,math.pi/2)
        epsilon = 2*(np.sin(c*U)/(np.cos(U)**(1/c)))*(np.cos(U-c*U)/W)**((1-c)/c)
        return epsilon 
        
    def data(self):
        X = random.random()
        Y = self.f_rho_cal(X)
        if self.noise == 'empty':
            Z = Y
        elif self.noise == 'uniform':
            Z = Y + random.uniform(-0.05,0.05)
        elif self.noise == 'gauss':
            Z = Y + random.gauss(0,0.5)
        elif self.noise == 'heavy-tail':
            Z = Y + self.data_noise() 
        return (X,Z,Y)
    
    def f_rho_cal(self,x):
        return  np.exp(0.7*np.cos(2*np.pi*x))*np.sin(2*np.pi*x) + 0.5*np.sin(6*np.pi*x) - 0.3*np.sin(10*np.pi*x) + 0.1*np.sin(14*np.pi*x) + 2*abs(np.cos(2*np.pi*x))**(7/2) - 0.5*abs(np.sin(2*np.pi*(x-0.25)))**(7/2)

    def fuhao(self):
        if self.p-2 < 0:
            b = 1
        else:
            b = 0
        return b
    
    def iteration(self,X,Y):
        if self.epoch**self.theta > self.Ln:
            self.Ln += 1
            self.Num = np.append(self.Num,self.Ln-1)
            self.fn = np.append(self.fn,0)
            self.f_average_n = np.append(self.f_average_n,0)
        self.gamma = self.gamma0*(self.epoch**(-self.tau))
        Output = self.cal_basis_function(X)
        fX = np.sum(self.fn*Output)
        w_f = self.q*(abs(fX-Y)**(self.q-1))*np.sign(fX-Y)
        J_list = np.array([ j**(2*self.s*self.p) for j in range(1,self.Ln+1)])
        B = (np.sum(J_list*(np.abs(self.fn)**self.p))+10**(-8))**(max(0,2-self.p)/self.p)
        Alpha = B*(np.abs(self.fn)**(self.p-1))*np.sign(self.fn)*J_list-self.gamma*w_f*Output
        J_list_new = np.array([ j**(-2*self.s*self.pp) for j in range(1,self.Ln+1)])
        A = (np.sum(J_list_new*(np.abs(Alpha)**self.pp))+10**(-8))**(min(0,self.p-2)/self.p)
        self.fn = A*J_list_new*(np.abs(Alpha)**(1/(self.p-1)))*np.sign(Alpha)
        if self.epoch >= self.sum_epoch/2:
            self.f_average_n = self.n/(self.n+1)*self.f_average_n + 1/(self.n+1)*self.fn
            self.n += 1
        #print(A,B)

    
    def gauss_legendre_01(self, n_x):
        """
        Gauss--Legendre nodes and weights on [0,1].
        """
        t, w = np.polynomial.legendre.leggauss(n_x)
        x = 0.5 * (t + 1.0)
        w = 0.5 * w
        return x, w


    def sample_abs_noise(self, n_noise):
        """
        Sample eta_k = |xi_k| from the noise distribution.

        Because self.data() returns (X, noisy_value, clean_value)
        after local assignment X, Y, Z = self.data(),
        the noise is Y - Z.
        """
        eta_list = []
        for _ in range(n_noise):
            if self.noise == 'empty':
                xi = 0
            elif self.noise == 'uniform':
                xi = random.uniform(-0.05,0.05)
            elif self.noise == 'gauss':
                xi = random.gauss(0,0.5)
            elif self.noise == 'heavy-tail':
                xi = self.data_noise()
            eta_list.append(abs(xi))
        return np.array(eta_list, dtype=float)


    def test_lq_gauss_symmetric(self, n_x=1024, n_noise=4096, chunk=1024):
        """
        Evaluate errors using:
        1. Gauss--Legendre quadrature for x in [0,1];
        2. symmetric noise pairing eta = |xi|.

        It computes both:

        clean Lq error:
            int_0^1 |f(x)-f_rho(x)|^q dx,

        and symmetric noisy excess risk:
            int_0^1 E_eta [
                (|d(x)-eta|^q + |d(x)+eta|^q)/2 - eta^q
            ] dx,

        where d(x)=f(x)-f_rho(x).
        """

        q = self.q

        # Gauss--Legendre points for x
        xs, ws = self.gauss_legendre_01(n_x)

        # Symmetric noise samples eta=|xi|
        eta = self.sample_abs_noise(n_noise)

        # Evaluate f, averaged f, and f_rho on Gauss nodes
        f_vals = np.array([self.fn_cal(x) for x in xs], dtype=float)
        fave_vals = np.array([self.f_average_n_cal(x) for x in xs], dtype=float)
        frho_vals = np.array([self.f_rho_cal(x) for x in xs], dtype=float)

        d_fn = f_vals - frho_vals
        d_ave = fave_vals - frho_vals

        # Clean Lq errors:
        # int |f-f_rho|^q dx
        clean_fn = np.sum(ws * np.abs(d_fn)**q)
        clean_ave = np.sum(ws * np.abs(d_ave)**q)

        # Symmetric noisy excess risks:
        # int E_eta A_q(d(x), eta) dx
        # A_q(d, eta) = (|d-eta|^q + |d+eta|^q)/2 - eta^q
        A_fn_sum = np.zeros(n_x, dtype=float)
        A_ave_sum = np.zeros(n_x, dtype=float)

        # Also compute noisy risks themselves if needed:
        # int E_eta (|d-eta|^q + |d+eta|^q)/2 dx
        risk_fn_sum = np.zeros(n_x, dtype=float)
        risk_ave_sum = np.zeros(n_x, dtype=float)

        d_fn_mat = d_fn.reshape(-1, 1)
        d_ave_mat = d_ave.reshape(-1, 1)

        for start in range(0, n_noise, chunk):
            end = min(start + chunk, n_noise)
            eta_chunk = eta[start:end].reshape(1, -1)

            eta_q = eta_chunk**q

            risk_fn_chunk = 0.5 * (
                np.abs(d_fn_mat - eta_chunk)**q
                +
                np.abs(d_fn_mat + eta_chunk)**q
            )

            risk_ave_chunk = 0.5 * (
                np.abs(d_ave_mat - eta_chunk)**q
                +
                np.abs(d_ave_mat + eta_chunk)**q
            )

            A_fn_chunk = risk_fn_chunk - eta_q
            A_ave_chunk = risk_ave_chunk - eta_q

            # Mathematically A_q >= 0.
            # This only removes tiny negative values caused by floating-point cancellation.
            A_fn_chunk = np.maximum(A_fn_chunk, 0.0)
            A_ave_chunk = np.maximum(A_ave_chunk, 0.0)

            risk_fn_sum += risk_fn_chunk.sum(axis=1)
            risk_ave_sum += risk_ave_chunk.sum(axis=1)

            A_fn_sum += A_fn_chunk.sum(axis=1)
            A_ave_sum += A_ave_chunk.sum(axis=1)

        risk_fn = np.sum(ws * (risk_fn_sum / n_noise))
        risk_ave = np.sum(ws * (risk_ave_sum / n_noise))

        excess_fn = np.sum(ws * (A_fn_sum / n_noise))
        excess_ave = np.sum(ws * (A_ave_sum / n_noise))

        '''
        print(
            'epoch:', self.epoch,
            'Ln:', self.Ln,
            'clean_fn:', clean_fn,
            'clean_ave:', clean_ave,
            'risk_fn:', risk_fn,
            'risk_ave:', risk_ave,
            'excess_fn:', excess_fn,
            'excess_ave:', excess_ave
        )'''

        return {
            'clean_fn': clean_fn,
            'clean_ave': clean_ave,
            'risk_fn': risk_fn,
            'risk_ave': risk_ave,
            'excess_fn': excess_fn,
            'excess_ave': excess_ave,
        }

    
    def algorithm(self,epoch):
        S_ave = 0
        S_estimator = 0
        self.sum_epoch = epoch
        self.n = 0
        for i in range(epoch):
            self.epoch = i+1
            X,Y,Z = self.data()
            self.iteration(X,Y)
        result = self.test_lq_gauss_symmetric(
            n_x=512,
                n_noise=1024,
            chunk=512
        )
        return (result['excess_ave'],result['excess_fn'],result['clean_ave'],result['clean_fn'])

```

```python
import math
import random
import matplotlib.pyplot as plt
import time
import numpy as np

class Mirror_descent_Banach():
    def __init__(self,noise,para,s,theta,gamma0,N,L=10.):
        self.s = s
        self.p = 1
        self.q = 1
        self.L = L
        self.N = N
        self.theta = theta
        self.Ln = math.ceil(self.N**self.theta)
        initial_alpha = (3*self.L/(math.pi**2)* np.array([ j**(-2*self.s-2) for j in range(1,self.Ln+1) ]))
        self.alpha_plus = initial_alpha.copy()
        self.alpha_minus = initial_alpha.copy()
        self.fn = self.alpha_plus-self.alpha_minus
        self.f_average_n = np.zeros(self.Ln)
        self.gamma0 = gamma0 #Initial step size
        self.gamma = self.gamma0*(self.N**(-1/2))
        self.Num = np.zeros(1)
        self.epoch = 1 #  Current number of iterations
        self.a0 = 2*np.cos(1.5*math.pi*np.arccos(1-abs(0-0.5))) + (2*0)**2 - 3*abs(0-1/4)**(2/3) + 3*np.cos(5*0) - np.exp(3*0) 
        self.a1 = 2*np.cos(1.5*math.pi*np.arccos(1-abs(1-0.5))) + (2*1)**2 - 3*abs(1-1/4)**(2/3) + 3*np.cos(5*1) - np.exp(3*1) 
        self.noise = noise
        self.para = para
        self.J_list = np.array([ j**(2*self.s) for j in range(1,self.Ln+1)])
        self.J_list_new = np.array([ j**(-2*self.s) for j in range(1,self.Ln+1)])
    
    def cal_basis_function(self,X):
        Output = franklin_system.franklin_system_values_fast(self.Ln, X)
        return Output
    
    def fn_cal(self,X):
        Output = self.cal_basis_function(X)
        return np.sum(self.fn*Output)
        
    def f_average_n_cal(self,X):
        Output = self.cal_basis_function(X)
        return np.sum(self.f_average_n*Output)

    def data_noise(self):
        c = self.para
        W = np.random.exponential()
        U = np.random.uniform(-math.pi/2,math.pi/2)
        epsilon = 2*(np.sin(c*U)/(np.cos(U)**(1/c)))*(np.cos(U-c*U)/W)**((1-c)/c)
        return epsilon 
        
    def data(self):
        X = random.random()
        Y = self.f_rho_cal(X)
        if self.noise == 'empty':
            Z = Y
        elif self.noise == 'uniform':
            Z = Y + random.uniform(-0.05,0.05)
        elif self.noise == 'gauss':
            Z = Y + random.gauss(0,0.5)
        elif self.noise == 'heavy-tail':
            Z = Y + self.data_noise() 
        return (X,Z,Y)
    
    def f_rho_cal(self,x):
        return  np.exp(0.7*np.cos(2*np.pi*x))*np.sin(2*np.pi*x) + 0.5*np.sin(6*np.pi*x) - 0.3*np.sin(10*np.pi*x) + 0.1*np.sin(14*np.pi*x) + 2*abs(np.cos(2*np.pi*x))**(7/2) - 0.5*abs(np.sin(2*np.pi*(x-0.25)))**(7/2)
        
    
    def iteration(self,X,Y):
        fX = self.fn_cal(X)
        w_f = np.sign(fX-Y)
        Output = self.cal_basis_function(X)
        Alpha_plus = self.alpha_plus*np.exp(-self.gamma*w_f*self.J_list_new*Output)
        Alpha_minus = self.alpha_minus*np.exp( self.gamma*w_f*self.J_list_new*Output)
        A = max(1, np.sum(self.J_list*(Alpha_plus+Alpha_minus))/self.L)
        self.alpha_plus = Alpha_plus/A
        self.alpha_minus = Alpha_minus/A
        self.fn = self.alpha_plus-self.alpha_minus
        if self.epoch >= self.sum_epoch/2:
            self.f_average_n = (self.n/(self.n+1)*self.f_average_n + 1/(self.n+1)*self.fn)
            self.n += 1

    
    def gauss_legendre_01(self, n_x):
        """
        Gauss--Legendre nodes and weights on [0,1].
        """
        t, w = np.polynomial.legendre.leggauss(n_x)
        x = 0.5 * (t + 1.0)
        w = 0.5 * w
        return x, w


    def sample_abs_noise(self, n_noise):
        """
        Sample eta_k = |xi_k| from the noise distribution.

        Because self.data() returns (X, noisy_value, clean_value)
        after local assignment X, Y, Z = self.data(),
        the noise is Y - Z.
        """
        eta_list = []
        for _ in range(n_noise):
            if self.noise == 'empty':
                xi = 0
            elif self.noise == 'uniform':
                xi = random.uniform(-0.05,0.05)
            elif self.noise == 'gauss':
                xi = random.gauss(0,0.5)
            elif self.noise == 'heavy-tail':
                xi = self.data_noise()
            eta_list.append(abs(xi))
        return np.array(eta_list, dtype=float)


    def test_lq_gauss_symmetric(self, n_x=1024, n_noise=4096, chunk=1024):
        """
        Evaluate errors using:
        1. Gauss--Legendre quadrature for x in [0,1];
        2. symmetric noise pairing eta = |xi|.

        It computes both:

        clean Lq error:
            int_0^1 |f(x)-f_rho(x)|^q dx,

        and symmetric noisy excess risk:
            int_0^1 E_eta [
                (|d(x)-eta|^q + |d(x)+eta|^q)/2 - eta^q
            ] dx,

        where d(x)=f(x)-f_rho(x).
        """

        q = self.q

        # Gauss--Legendre points for x
        xs, ws = self.gauss_legendre_01(n_x)

        # Symmetric noise samples eta=|xi|
        eta = self.sample_abs_noise(n_noise)

        # Evaluate f, averaged f, and f_rho on Gauss nodes
        f_vals = np.array([self.fn_cal(x) for x in xs], dtype=float)
        fave_vals = np.array([self.f_average_n_cal(x) for x in xs], dtype=float)
        frho_vals = np.array([self.f_rho_cal(x) for x in xs], dtype=float)

        d_fn = f_vals - frho_vals
        d_ave = fave_vals - frho_vals

        # Clean Lq errors:
        # int |f-f_rho|^q dx
        clean_fn = np.sum(ws * np.abs(d_fn)**q)
        clean_ave = np.sum(ws * np.abs(d_ave)**q)

        # Symmetric noisy excess risks:
        # int E_eta A_q(d(x), eta) dx
        # A_q(d, eta) = (|d-eta|^q + |d+eta|^q)/2 - eta^q
        A_fn_sum = np.zeros(n_x, dtype=float)
        A_ave_sum = np.zeros(n_x, dtype=float)

        # Also compute noisy risks themselves if needed:
        # int E_eta (|d-eta|^q + |d+eta|^q)/2 dx
        risk_fn_sum = np.zeros(n_x, dtype=float)
        risk_ave_sum = np.zeros(n_x, dtype=float)

        d_fn_mat = d_fn.reshape(-1, 1)
        d_ave_mat = d_ave.reshape(-1, 1)

        for start in range(0, n_noise, chunk):
            end = min(start + chunk, n_noise)
            eta_chunk = eta[start:end].reshape(1, -1)

            eta_q = eta_chunk**q

            risk_fn_chunk = 0.5 * (
                np.abs(d_fn_mat - eta_chunk)**q
                +
                np.abs(d_fn_mat + eta_chunk)**q
            )

            risk_ave_chunk = 0.5 * (
                np.abs(d_ave_mat - eta_chunk)**q
                +
                np.abs(d_ave_mat + eta_chunk)**q
            )

            A_fn_chunk = risk_fn_chunk - eta_q
            A_ave_chunk = risk_ave_chunk - eta_q

            # Mathematically A_q >= 0.
            # This only removes tiny negative values caused by floating-point cancellation.
            A_fn_chunk = np.maximum(A_fn_chunk, 0.0)
            A_ave_chunk = np.maximum(A_ave_chunk, 0.0)

            risk_fn_sum += risk_fn_chunk.sum(axis=1)
            risk_ave_sum += risk_ave_chunk.sum(axis=1)

            A_fn_sum += A_fn_chunk.sum(axis=1)
            A_ave_sum += A_ave_chunk.sum(axis=1)

        risk_fn = np.sum(ws * (risk_fn_sum / n_noise))
        risk_ave = np.sum(ws * (risk_ave_sum / n_noise))

        excess_fn = np.sum(ws * (A_fn_sum / n_noise))
        excess_ave = np.sum(ws * (A_ave_sum / n_noise))

        '''
        print(
            'epoch:', self.epoch,
            'Ln:', self.Ln,
            'clean_fn:', clean_fn,
            'clean_ave:', clean_ave,
            'risk_fn:', risk_fn,
            'risk_ave:', risk_ave,
            'excess_fn:', excess_fn,
            'excess_ave:', excess_ave
        )'''

        return {
            'clean_fn': clean_fn,
            'clean_ave': clean_ave,
            'risk_fn': risk_fn,
            'risk_ave': risk_ave,
            'excess_fn': excess_fn,
            'excess_ave': excess_ave,
        }

    
    def algorithm(self):
        self.sum_epoch = self.N + 1
        self.n = 0
        for i in range(1,self.sum_epoch):
            self.epoch = i
            X,Y,Z = self.data()
            self.iteration(X,Y)
        result = self.test_lq_gauss_symmetric(
            n_x=512,
                n_noise=1024,
            chunk=512
        )
        return (result['excess_ave'],result['excess_fn'],result['clean_ave'],result['clean_fn'])

```
